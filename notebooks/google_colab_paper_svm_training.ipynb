# Google Colab Training Notebook for Paper-Based SVM Models

This notebook trains the paper-based smoke detection SVMs in Google Colab and saves the exported bundle as `paper_svm_models.pkl` into Google Drive so you can copy it into `backend/weights/` in your local repository.

## What this notebook does

1. Installs the required Python packages.
2. Mounts Google Drive.
3. Points Colab at your repository files and labeled training data.
4. Imports the existing training script from the repo.
5. Trains the SVM voting classifier.
6. Saves `paper_svm_models.pkl` to Drive.
7. Verifies the file exists so you can download it to your local `backend/weights/` folder.
